#EDA analysis

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("/content/zomato.csv", encoding='latin1')
df.head()
```

```python
df.describe()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
#Duplicates

df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='Price range')
plt.xlabel("Price range")
plt.ylabel("Count")
plt.title("Distribution of Price Ranges")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Price range', y='Aggregate rating')
plt.title("Price Range vs. Aggregate Rating")
plt.xlabel("Price Range")
plt.ylabel("Aggregate Rating")
plt.show()
#
```

```python
plt.figure(figsize=(10,6))
sns.barplot(data=df, x='Votes', y='Aggregate rating')
plt.title("Votes vs. Aggregate Rating")
plt.xlabel("Votes")
plt.ylabel("Aggregate Rating")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x='Votes', y='Rating color')
plt.title("Votes vs. Rating Color")
plt.xlabel("Votes")
plt.ylabel("Rating Color")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(data=df, x='Votes', y='Rating text')
plt.title("Votes vs. Rating Text")
plt.xlabel("Votes")
plt.ylabel("Rating Text")
plt.show()
```

```python
cuisin_counts = df['Cuisines'].value_counts().head(10)
plt.figure(figsize=(10,6))
cuisin_counts.plot(kind="bar")
plt.title("Top 10 Cuisines")
plt.xlabel("Cuisines")
plt.xticks(rotation=45)
plt.ylabel("Count")
plt.show()
```
